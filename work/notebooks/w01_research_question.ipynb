## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

- **Search question:** Which pages should a content team review first for refresh, expansion, protection, pruning or monitoring?
- **Unit of analysis:** one page (content item).
- **Output:** a ranked review queue with a score, a suggested action and reason codes for each page.
- **Decision:** which pages get the team's limited review time this cycle.
- **Action:** a reviewer checks the top pages and decides whether to refresh, expand, protect, prune or just monitor them.
- **Cost of a wrong recommendation:** a false positive wastes reviewer time on a healthy page; a missed page keeps losing visibility unnoticed. Because reviewers can only check a limited number of pages, the top of the list matters most, so I will judge quality with Precision@K.
- **Why data/ML can help:** a team cannot inspect thousands of pages by hand, and signals such as impressions, position, CTR, page age and trend can be combined and checked against outcomes. A transparent baseline score can be compared with a learned model.
- **Why this is not just "train a model":** the deliverable is a ranked, explainable action queue with reason codes. "Right page to fix" means the right page to review first, not a guaranteed recovery.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

**Numbers from the starter data: 30,000 pages after the starter filters, of which 54.2% are labelled 'down' (median impressions_90d = 731). Average CTR falls sharply with position, from about 2.71 in positions 0-3 to about 0.65 (3-10), 0.32 (10-20) and 0.21 (20+). This suggests there is a lot of room for a ranked review queue, and it is an observed, directional pattern in this sample, not proof of how Google works.**

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import os
import pandas as pd

LOCAL = "data/raw/content_refresh_anonymized.csv"
URL = "https://raw.githubusercontent.com/Atul200429/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(LOCAL if os.path.exists(LOCAL) else URL)
d = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)].drop_duplicates("content_id")

print("Rows in starter file:", len(df))
print("Rows after starter filters (impressions_90d > 0, age >= 90):", len(d))
print("Share with trend_direction == 'down':", round((d["trend_direction"] == "down").mean(), 3))
print("Median impressions_90d:", d["impressions_90d"].median())
print(d.groupby(pd.cut(d["avg_position"], [0, 3, 10, 20, 100]), observed=False)["ctr"].mean())
```
